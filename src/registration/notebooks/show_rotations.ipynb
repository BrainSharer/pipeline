```python
import numpy as np
import matplotlib.pyplot as plt
import tifffile
```

```python
# orig is z,y,x (11, 87, 54)
# X, cols should be biggest = 87
# Y, rows should be smallest = 11
# Z, depth should be 54
# vol will be 54,11,87
volpath = "/net/birdstore/Active_Atlas_Data/data_root/pipeline_data/CTB016/preps/testing.tif"
arr3d = tifffile.imread(volpath)
print('pre shape: z,y,x', arr3d.shape)
# 3. Rotate the array by 90 degrees along specific axes
# Axes (0, 1) means we rotate the 0th axis (Z) toward the 1st axis (Y)
```

```python
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.voxels(arr3d, edgecolor='k')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.show()
```

```python
rot1 = np.rot90(arr3d, k=1, axes=(0,1))
print('post shape: ', rot1.shape)
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.voxels(rot1, edgecolor='k')
ax.set_xlabel('X Axis')
ax.set_ylabel('Y Axis')
ax.set_zlabel('Z Axis')
plt.show()
```
